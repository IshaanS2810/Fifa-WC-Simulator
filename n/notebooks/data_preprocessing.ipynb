# Data Preprocessing

This notebook will document the raw dataset loading, cleaning, and standardization steps for FIFA World Cup simulation data.

```python
import pandas as pd

matches = pd.read_csv("../datasets/raw/results.csv")
matches.head()
matches.info()
matches.isnull().sum()
matches["date"] = pd.to_datetime(matches["date"])
matches = matches[matches["date"] >= "1990-01-01"]
important_tournaments = [
    "FIFA World Cup",
    "UEFA Euro",
    "Copa America",
    "UEFA Nations League",
    "African Cup of Nations",
    "AFC Asian Cup",
    "FIFA World Cup qualification"
]

matches = matches[
    matches["tournament"].isin(important_tournaments)
]
TEAM_MAP = {
    "USA": "United States",
    "Korea Republic": "South Korea",
    "IR Iran": "Iran"
}

matches["home_team"] = matches["home_team"].replace(TEAM_MAP)
matches["away_team"] = matches["away_team"].replace(TEAM_MAP)
matches = matches.drop_duplicates()
matches.to_csv(
    "../datasets/cleaned/cleaned_matches.csv",
    index=False
)
```

```python
import pandas as pd
players = pd.read_csv("../datasets/raw/players.csv")
players.head()
players.info()
players.isnull().sum()
important_player_cols = [
    "player_id",
    "name",
    "country_of_citizenship",
    "position",
    "market_value_in_eur",
    "date_of_birth"
]
players = players[important_player_cols].copy()

players = players.rename(columns={
    "country_of_citizenship": "nationality"
})

players = players.dropna(
    subset=["player_id", "name", "nationality"]
)
players.to_csv(
    "../datasets/cleaned/cleaned_players.csv",
    index=False
)
```

```python
appearances = pd.read_csv("../datasets/raw/appearances.csv")
appearances.head()
appearances.info()
important_appearance_cols = [
    "player_id",
    "game_id",
    "minutes_played",
    "goals",
    "assists",
    "yellow_cards",
    "red_cards"
]

appearances = appearances[important_appearance_cols]
appearances = appearances.fillna(0)
appearances.to_csv(
    "../datasets/cleaned/cleaned_appearances.csv",
    index=False
)
```

```python
import pandas as pd

fifa_rankings = pd.read_csv(
    "../datasets/raw/fifa_ranking-2024-06-20.csv"
)
```

```python
fifa_rankings = fifa_rankings[
    [
        "rank_date",
        "country_full",
        "rank",
        "total_points"
    ]
].copy()
```

```python
fifa_rankings = fifa_rankings.rename(
    columns={
        "rank_date": "date",
        "country_full": "nationality",
        "rank": "fifa_rank",
        "total_points": "fifa_points"
    }
)
```

```python
fifa_rankings["date"] = pd.to_datetime(
    fifa_rankings["date"]
)
```

```python
fifa_rankings = fifa_rankings.drop_duplicates(
    subset=[
        "date",
        "nationality"
    ]
)
fifa_rankings = fifa_rankings.sort_values(
    [
        "nationality",
        "date"
    ]
).reset_index(
    drop=True
)
```

```python
country_mapping = {

    "United States": "USA",
    "Korea Republic": "South Korea",
    "IR Iran": "Iran",
    "Czechia": "Czech Republic",
    "China PR": "China",
    "Republic of Ireland": "Ireland",
    "Congo DR": "DR Congo",
    "Cape Verde Islands": "Cape Verde",
    "FYR Macedonia": "North Macedonia"
}
fifa_rankings["nationality"] = (
    fifa_rankings["nationality"]
    .replace(country_mapping)
)
```

```python
fifa_rankings.to_csv(
    "../datasets/cleaned/cleaned_fifa_rankings.csv",
    index=False
)
```
